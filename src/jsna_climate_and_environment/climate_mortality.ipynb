```python
import polars as pl
import polars.selectors as cs
import plotly.express as px
import fsspec
import fastexcel
from loguru import logger
```

```python
def read_excel_tables(url) -> fastexcel.ExcelReader:
    with fsspec.open(url) as file:
        logger.info(f"reading excell file from {url}")
        reader = fastexcel.read_excel(file.read())
    logger.info(f"file contains the sheets: {reader.sheet_names}")
    return reader

reader = read_excel_tables("https://assets.publishing.service.gov.uk/media/67ebd7ade9c76fa33048c592/heat-mortality-monitoring-data-2024.ods")
```
