# The tableau data model will minimize data duplication and keep dashboards small

The data model will require:
- each area polygon for the current year
- the metadata for each indicator
- the demographic breakdown for each indicator
- the time_range for each indicator

indicator_fact
- id
- area_id
- metadata_id
- population_id
- time_range_id
- value
- error_margin

area_dim
- id
- name
- type
- source_url
- parent_district: str | None
- parent_lgr: str | None
- geometry

metadata_dim
- id
- name
- description
- source_system
- source_identifier | optional
- type_of_value
- max_value
- min_value
- missing_value_description

population_dim
- population_id
- name
- description
- source
- date_recorded
- count
- date_updated
